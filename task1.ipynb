```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

img = np.array(Image.open('img.jpeg').convert('L'))

factors = [1, 2, 4, 8]

plt.figure(figsize=(12, 8))

for i, factor in enumerate(factors):
    sampled = img[::factor, ::factor]
    plt.subplot(2, 2, i + 1)
    plt.imshow(sampled, cmap='gray')
    plt.title(f'Sampling Factor = {factor}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

img = np.array(Image.open('img.jpeg').convert('L'))

levels = [256, 64, 16, 4]

plt.figure(figsize=(12, 8))

for i, L in enumerate(levels):
    quantized = np.floor(img / 256 * L) * (256 / L)
    plt.subplot(2, 2, i + 1)
    plt.imshow(quantized.astype(np.uint8), cmap='gray')
    plt.title(f'Quantization Levels = {L}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
