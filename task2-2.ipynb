```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

img = np.array(Image.open('img.jpeg').convert('L'))
img2 = np.array(Image.open('img2.jpeg').convert('L'))

img2 = np.array(Image.fromarray(img2).resize((img.shape[1], img.shape[0])))

result = np.abs(img.astype(int) - img2.astype(int))

plt.imshow(result, cmap='gray')
plt.title('Subtraction')
plt.axis('off')
plt.show()
```
